```python
import numpy as np

STUDENT_NAME = "Adam Niegel S. Aspiras"
STUDENT_ID = int("TUPM-25-0389".split('-')[2]) # Result: 389

# Set seed for consistent random results
np.random.seed(STUDENT_ID)

# Generate 10 random scores
# Range: 0 to (389 % 100 + 50) = 139
scores = np.random.randint(0, STUDENT_ID % 100 + 50, 10)

# Vectorized conditional logic
# Threshold: 389 % 70 = 39
grades = np.where(scores >= STUDENT_ID % 70, 'Pass', 'Fail')

print(f"{STUDENT_NAME}'s Scores:", scores)
print("Grades:", grades)
```
